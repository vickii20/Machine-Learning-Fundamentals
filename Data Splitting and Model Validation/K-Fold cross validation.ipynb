# Function - (K-Fold cross validation):

Data Split Configuration :
  1.shuffle

```python
import pandas as pd
from sklearn.model_selection import train_test_split, KFold, cross_val_score
```

```python
df = pd.read_csv(r"E:\Machine Learning\Data Splitting and Model Validation\Sources\dataset.csv")
```

```python
X = df[['age', 'experience_years', 'education_level', 'hours_per_week']] 
```

```python
# behind the formula of - ** n_split **

# Test size  = Total samples / n_splits
# Train size = Total samples − Test size

# Test = 10000 / 5 = 2000
# Train = 10000 − 2000 = 8000

kf = KFold(
    n_splits = 5,
    shuffle = True,
    random_state= 42
)   
```

```python
z = 0
for train_idx, test_idx in kf.split(X):
    print(f"iteration is {z}")
    print(train_idx, test_idx)
    z = z + 1
    print('#####')
```
