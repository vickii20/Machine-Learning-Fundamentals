# Training Phase - 1 : # for salary prediction (Function - train_test_split)

```python
# train_test_split :
```

```python
X = df[['age', 'experience_years', 'education_level', 'hours_per_week']]
Y = df[['salary']]
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    X,
    Y,
    test_size = 0.2,
    random_state = 42
)
```

```python
model = sm.OLS(y_train, x_train)
results = model.fit()
```
