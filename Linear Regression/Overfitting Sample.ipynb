```python
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score
```

```python
import pandas as pd
```

```python
df = pd.read_csv(r"E:\Machine Learning\Linear Regression\source\overfitting_demo_10000x201.csv")
```

```python
df.head()
```

```python
len(df)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop('target', axis = 1)
Y = df['target']

x_train, x_test, y_train, y_test = train_test_split(
    X,
    Y,
    train_size = 0.08, # 8% for overfitting
    random_state = 42
)
```

```python
ols = Pipeline([
    ('scaler', StandardScaler()),
    ('ols', LinearRegression())
]
)
```

```python
# training with train datasets
ols.fit(x_train,y_train)
```

```python
# predict first with same train dataset to check wheather it learn correctly from the train dataset
train_pred = ols.predict(x_train)
# this line passing the model with new datasets
test_pred = ols.predict(x_test)
```

```python
# cost function
# MSE

train_mse = mean_squared_error(y_train, train_pred)
test_mse = mean_squared_error(y_test, test_pred)

# R-squared

train_r2 = r2_score(y_train, train_pred)
test_r2 = r2_score(y_test, test_pred)
```

```python
print(f"train_mse - {train_mse}\ntest_mse - {test_mse}\ntrain_r2 - {train_r2}\ntest_r2 - {test_r2}")


# this model is overfitting - low bias, high variance
```

```python
# rigde regression - to reduce the large weights of the regression line...
```

```python
from sklearn.linear_model import Ridge
```

```python
ridge = Pipeline([
    ('scaler', StandardScaler()),
    ('ridge', Ridge(alpha=50))
])
```

```python
ridge.fit(x_train, y_train)
```

```python
train_pred_rid = ridge.predict(x_train)
test_pred_rid = ridge.predict(x_test)
```

```python
train_mse_rid = mean_squared_error(y_train, train_pred_rid)
test_mse_rid = mean_squared_error(y_test, test_pred_rid)

train_r2_rid = r2_score(y_train, train_pred_rid)
test_r2_rid = r2_score(y_test, test_pred_rid)
```

```python
print('///ridge//')
print(f"Ridge Train MSE: {train_mse_rid}\nRidge Test MSE: {test_mse_rid}\nRidge Train R2: {train_r2_rid}\nRidge Test R2: {test_r2_rid}")
```

```python
print('//linear regression - overfitting///')
print(f"train_mse - {train_mse}\ntest_mse - {test_mse}\ntrain_r2 - {train_r2}\ntest_r2 - {test_r2}")
```
